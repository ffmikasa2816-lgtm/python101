## 🔮 Function คืออะไร?
ลองนึกว่า Function คือ 'สูตรเวทมนตร์' ที่เราเก็บไว้เรียกใช้ซ้ำได้
เมื่อเรียกชื่อสูตรนั้น โปรแกรมจะทำตามที่เรากำหนดไว้ทันที

```python
# ตัวอย่าง: เสกเสียงหมา
# no input
def dog():
    print("โฮ่งๆ 🐶")

# เรียกใช้ฟังก์ชัน
print("เรียกหมา!")
dog()
dog()
dog()
```

## 🌟 ฟังก์ชันทำนายอนาคต (Fortune Teller)

```python
import random

def ทำนายอนาคต():
    คำทำนาย = [
        "วันนี้จะมีเรื่องดีๆ เกิดขึ้นแบบไม่คาดคิด",
        "ระวังคำพูด อาจทำให้เข้าใจผิด",
        "คุณจะได้รับข่าวดีเรื่องงาน",
        "มีคนแอบชอบอยู่เงียบๆ",
        "ได้เงินจากทางที่ไม่คาดคิด",
        "ลองเสี่ยงโชคดูสิ อาจจะเฮง!"
    ]
    return random.choice(คำทำนาย)

print("🔮 คำทำนายของคุณคือ:")
print(ทำนายอนาคต())
```

## 🍜 ฟังก์ชันเลือกเมนูอาหารเย็นแบบสุ่ม

```python
import random

def เมนูเย็นนี้():
    เมนู = [
        "ข้าวผัดกะเพราไข่ดาว",
        "ข้าวมันไก่",
        "หมูกระทะกับเพื่อนๆ",
        "สุกี้แห้งทะเล",
        "ตำถาดแซ่บๆ",
        "ราเมนญี่ปุ่น",
        "ข้าวไข่ข้นแฮมชีส",
        "แกงเขียวหวานไก่"
    ]
    return random.choice(เมนู)

print("🍽️ เมนูเย็นนี้คือ:")
print(เมนูเย็นนี้())
```

# รวมเป็นแอปพยากรณ์โชคชะตาแบบโต้ตอบ



```python
def start_app():
    print("🎲 โปรแกรมพยากรณ์โชคชะตา")
    print("1. ทำนายอนาคต")
    print("2. แนะนำเมนูอาหารเย็น")
    choice = input("คุณอยากเลือกข้อไหน (1/2): ")

    if choice == "1":
        print("🔮 คำทำนาย:")
        print(ทำนายอนาคต())
    elif choice == "2":
        print("🍛 เมนูที่น่ากินคืนนี้:")
        print(เมนูเย็นนี้())
    else:
        print("กรุณาเลือกแค่ 1 หรือ 2 นะจ๊ะ~")

start_app()
```

## Creative Idea


*   นำเสนอ idea การนำ function ไปใช้แก้ปัญหาในชีวิตประจำวัน เช่น วันนี้กินอะไรดีน๊า, มื้อนี้ใครจ่าย
*   ยกตัวอย่าง code ให้เพื่อนๆดู




```python
def should_i_go(location, pm25_level, homework_items, budget):
    print(f"กำลังประเมินคำชวนไป {location}...")

    # เงื่อนไขที่ 1: สุขภาพมาก่อน
    if pm25_level > 100:
        return "❌ อย่าเพิ่งไปเลย ฝุ่นเชียงใหม่วันนี้หนักมาก พักปอดก่อนดีกว่า"

    # เงื่อนไขที่ 2: หน้าที่ต้องมาก่อน
    if homework_items >= 3:
        return "❌ การบ้านท่วมหัวแล้ว! ปั่นงานให้เสร็จก่อนเถอะหัวหน้า"

    # เงื่อนไขที่ 3: เงินในกระเป๋า
    if budget < 300 and location == "นิมมาน":
        return "⚠️ ไปได้นะ แต่ระวังงบบานปลาย แนะนำให้กินข้าวไข่เจียวก่อนไป"

    return "✅ ลุยเลย! ทำงานมาเหนื่อยแล้ว ให้รางวัลตัวเองบ้าง"

# ลองใช้งานฟังก์ชัน
decision = should_i_go(location="นิมมาน", pm25_level=85, homework_items=1, budget=500)
print(decision)
```
